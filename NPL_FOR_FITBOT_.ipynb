Understanding NLP Techniques
Before implementation:

Tokenization: Breaking text into individual words or sentences

Stopword Removal: Filtering out common words that don't add meaning (e.g., "the", "and")

Stemming: Crudely chopping off word endings (e.g., "running" → "run")

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

# Download required NLTK data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')

def process_text(text):
    print("Original Text:\n", text)

    # Tokenization
    words = word_tokenize(text)
    sentences = sent_tokenize(text)
    print("\nTokenized Words:", words)
    print("\nTokenized Sentences:", sentences)

    # Stopword Removal
    stop_words = set(stopwords.words('english'))
    filtered_words = [word for word in words if word.lower() not in stop_words and word.isalpha()]
    print("\nAfter Stopword Removal:", filtered_words)

    # POS Tagging
    tagged_words = pos_tag(filtered_words)
    print("\nPOS Tags:", tagged_words)

    # Stemming
    stemmer = PorterStemmer()
    stemmed_words = [stemmer.stem(word) for word in filtered_words]
    print("\nStemmed Words:", stemmed_words)

    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
    print("\nLemmatized Words:", lemmatized_words)

    return {
        'words': words,
        'sentences': sentences,
        'filtered_words': filtered_words,
        'tagged_words': tagged_words,
        'stemmed_words': stemmed_words,
        'lemmatized_words': lemmatized_words
    }

# Fitness-related text for FitBot
fitness_text = """
FitBot can help you achieve your fitness goals.
Running 3 times a week and doing strength training exercises
like push-ups and squats will improve your endurance and muscle strength.
Remember to stay hydrated and eat balanced meals for optimal results.
"""

# Process the text
processed_data = process_text(fitness_text)
```

```python
from nltk.corpus import reuters
import pandas as pd
from sklearn.model_selection import train_test_split

# For demonstration, we'll use a sample fitness dataset
# In practice, you'd want a real fitness Q&A dataset
data = {
    'text': [
        "How many calories should I eat to lose weight?",
        "What's the best exercise for building chest muscles?",
        "How often should I do cardio for heart health?",
        "Is weight training better than cardio for fat loss?",
        "What are good pre-workout meals?"
    ],
    'label': [
        "nutrition",
        "strength_training",
        "cardio",
        "comparison",
        "nutrition"
    ]
}

df = pd.DataFrame(data)

def preprocess_text(text):
    # Tokenize
    words = word_tokenize(text.lower())
    # Remove stopwords and non-alphabetic tokens
    words = [word for word in words if word.isalpha() and word not in stopwords.words('english')]
    # Lemmatize
    lemmatizer = WordNetLemmatizer()
    words = [lemmatizer.lemmatize(word) for word in words]
    return ' '.join(words)

# Apply preprocessing
df['processed_text'] = df['text'].apply(preprocess_text)
print("\nOriginal vs Processed Text:")
print(df[['text', 'processed_text']])
```

### 📖 Step 1: Understanding the Significance of Hyperparameter Tuning

Hyperparameters are external settings for a machine learning model that are not learned from the data.
They control the behavior of the training algorithm and the structure of the model itself.

Examples:
- For Decision Trees: `criterion`, `max_depth`, `min_samples_split`
- For Naive Bayes: `var_smoothing`

By tuning these hyperparameters, we can:
- Improve model performance
- Avoid overfitting or underfitting
- Optimize the model for better predictions

### 🌳 Step 3: Training and Evaluating Initial Decision Tree

We first train a Decision Tree without any hyperparameter tuning and evaluate its accuracy and confusion matrix.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# Load your dataset
df = pd.read_csv('gym_members_exercise_tracking.csv')

# Define features and target
X = df.drop('Workout_Type', axis=1)  # Example: Predict workout type
y = df['Workout_Type']

# Encode target if it's categorical
le = LabelEncoder()
y = le.fit_transform(y)

# ----> Create a OneHotEncoder object
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore') # sparse=False for dense output

# ----> Fit and transform the 'Gender' column
encoded_gender = encoder.fit_transform(X[['Gender']])

# ----> Create a DataFrame from the encoded data
encoded_gender_df = pd.DataFrame(encoded_gender, columns=encoder.get_feature_names_out(['Gender']))

# ----> Drop the original 'Gender' column and concatenate the encoded columns
X = X.drop('Gender', axis=1)
X = pd.concat([X, encoded_gender_df], axis=1)


# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train initial Decision Tree
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)

# Evaluation
print(f"🔵 Initial Decision Tree Accuracy: {accuracy_score(y_test, y_pred_dt)}")
print("🔵 Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))
```

### 🛠️ Step 4: Tuning the Decision Tree

We perform Grid Search with cross-validation to find the best parameters for Decision Tree.
Parameters to tune:
- `criterion`: 'gini' or 'entropy'
- `max_depth`: various depths
- `min_samples_split`: minimum number of samples required to split a node


```python
from sklearn.model_selection import GridSearchCV

# Define parameter grid
param_grid_dt = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [3, 5, 10, None],
    'min_samples_split': [2, 5, 10]
}

# Grid Search
grid_dt = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid_dt, cv=5)
grid_dt.fit(X_train, y_train)

# Best model
best_dt = grid_dt.best_estimator_
y_pred_best_dt = best_dt.predict(X_test)

# Evaluation
print(f"🔵 Best Decision Tree Parameters: {grid_dt.best_params_}")
print(f"🔵 Tuned Decision Tree Accuracy: {accuracy_score(y_test, y_pred_best_dt)}")
print("🔵 Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_best_dt))
```

### 🧠 Step 5: Training and Evaluating Initial Naive Bayes

We train a Gaussian Naive Bayes model without tuning and evaluate it.


```python
from sklearn.naive_bayes import GaussianNB

# Train initial Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
y_pred_nb = nb.predict(X_test)

# Evaluation
print(f"🟢 Initial Naive Bayes Accuracy: {accuracy_score(y_test, y_pred_nb)}")
print("🟢 Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_nb))
```

### 🛠️ Step 6: Tuning the Naive Bayes Model

For Naive Bayes, we tune the `var_smoothing` parameter, which controls numerical stability.


```python
# Define parameter grid
param_grid_nb = {
    'var_smoothing': [1e-9, 1e-8, 1e-7, 1e-6]
}

# Grid Search
grid_nb = GridSearchCV(GaussianNB(), param_grid_nb, cv=5)
grid_nb.fit(X_train, y_train)

# Best model
best_nb = grid_nb.best_estimator_
y_pred_best_nb = best_nb.predict(X_test)

# Evaluation
print(f"🟢 Best Naive Bayes Parameters: {grid_nb.best_params_}")
print(f"🟢 Tuned Naive Bayes Accuracy: {accuracy_score(y_test, y_pred_best_nb)}")
print("🟢 Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_best_nb))
/
```
